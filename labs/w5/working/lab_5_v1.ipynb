```python
import pandas as pd

cols = ["age","workclass","fnlwgt","education","education_num","marital_status",
        "occupation","relationship","race","gender","capital_gain","capital_loss",
        "hours_per_week","native_country","income_bracket"]

df = pd.read_csv("adult.data", names=cols, skipinitialspace=True, na_values="?")
# df = pd.read_csv("adult.data", names=cols, skipinitialspace=True)
print(df.shape)   # (32561, 15)
```

```python
# quick look at data to see if its clean, no duplicates, null values unusual characters etc
# convert it to d first so can easily run some checks over it
# df=pd.DataFrame(processed_per_file_data)
# view various dataframe details
# checking for null values and specific unknown values
print ("Rows     : " , df.shape[0])
print ("Columns  : " , df.shape[1])
print('------')
print ("\nFeatures : \n" , df.columns.tolist())
print('------')
print ("\nMissing values :  ", df.isnull().sum().values.sum())
print('------')
print ("\nMissing values per column:  ", df.isnull().sum())
print('------')
print('\nnum of dups in df   =   {}'.format(df.duplicated().sum()))
print('------')
print ("\nUnique values :  \n", df.nunique())
print('------')
print('\n null values = ', df.isnull().values.any())
print('------')
df_missing = (df=='?').sum()
print('\"  ?   \" values present =  \n',df_missing)
print('------')
df_unknown = (df=='unknown').sum()
print('\"   unknown   \" values present =  \n',df_unknown)
print('------')
df_na = df.isin(['N/A', 'N\\A', 'NA', 'n/a', 'n\\A', 'na', 'N_A']).sum()
print('    NA     values present = \n', df_na)
print('------')
df_none = df.isin(['None', 'none']).sum()
print('    none    values present = \n', df_none)
print('------')
# regex https://stackoverflow.com/questions/2124010/grep-regex-to-match-non-ascii-characters
# count of rows per column containing non-ascii characters
rogue_ascii_df = df.apply(lambda s: s.astype(str).str.contains(r"[^\x00-\x7F]", regex=True, na=False).sum())
print("    non ascii chars   values present:\n", rogue_ascii_df)
```
